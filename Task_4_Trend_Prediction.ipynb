Step 2: Analyze yearly content trends

```python
import pandas as pd
import matplotlib.pyplot as plt

# Load clean dataset
df = pd.read_csv("clean_netflix.csv")

# Count content released in each year
yearly_content = df["release_year"].value_counts().sort_index()

# Create line graph
plt.figure(figsize=(12, 6))

plt.plot(
    yearly_content.index,
    yearly_content.values,
    marker="o"
)

plt.title("Netflix Content Trend by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")
plt.grid(True)

plt.show()
```

Task 4 — Step 3: Build Forecasting Model

```python
from sklearn.linear_model import LinearRegression
import pandas as pd

# Prepare data
X = yearly_content.index.values.reshape(-1, 1)
y = yearly_content.values

# Create and train model
model = LinearRegression()
model.fit(X, y)

print("Model trained successfully!")
```

```python
# Future years
future_years = [[2022], [2023], [2024], [2025], [2026]]

# Make predictions
predictions = model.predict(future_years)

# Show predictions
for year, prediction in zip(range(2022, 2027), predictions):
    print(year, round(prediction))
```

Step 4: Visualize Future Predictions

```python
import matplotlib.pyplot as plt

future_years = [2022, 2023, 2024, 2025, 2026]

plt.figure(figsize=(12, 6))

# Historical data
plt.plot(
    yearly_content.index,
    yearly_content.values,
    label="Historical"
)

# Future predictions
plt.plot(
    future_years,
    predictions,
    marker="o",
    linestyle="--",
    label="Predicted"
)

plt.title("Netflix Content Trend: Historical and Future Predictions")
plt.xlabel("Year")
plt.ylabel("Number of Titles")
plt.legend()
plt.grid(True)

plt.show()
```

5. Interpret Results

### **5. Interpret Results**

* Netflix content increased significantly after the 2000s.
* The highest number of titles was recorded in **2018, with 1,146 titles**.
* After 2018, the number of titles decreased, reaching **592 titles in 2021**.
* The Linear Regression model predicts a slight increase from **383 titles in 2022** to **410 titles in 2026**.
* The predictions are lower than recent historical values, showing that the simple Linear Regression model has limitations for long-term forecasting.
