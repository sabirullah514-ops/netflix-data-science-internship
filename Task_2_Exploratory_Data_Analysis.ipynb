Step 1: Analyze dataset structure and statistics

```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("clean_netflix.csv")
```

```python
print("Shape:", df.shape)

print("\nColumns:")
print(df.columns.tolist())
```

```python
print(df.describe())
```

Step 2: Study content distribution by type.

```python
print(df["type"].value_counts())
```

Step 3: Identify Top Countries and Categories

```python
df["country"].head()
```

```python
df["country"].value_counts()
```

```python
countries = df["country"].str.split(", ").explode()

print(countries.value_counts().head(10))
```

```python
print(df["listed_in"].head())
```

```python
categories = df["listed_in"].str.split(", ").explode()

print(categories.value_counts().head(10))
```

```python
df["type"].value_counts().plot(kind="bar")

plt.title("Netflix Content by Type")
plt.xlabel("Content Type")
plt.ylabel("Number of Titles")
plt.show()
```

```python
top_countries = countries.value_counts().head(10)

top_countries.plot(kind="bar")

plt.title("Top 10 Countries by Netflix Content")
plt.xlabel("Country")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

```python
top_categories = categories.value_counts().head(10)

top_categories.plot(kind="bar")

plt.title("Top 10 Netflix Categories")
plt.xlabel("Category")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)

plt.show()
```

```python
year_counts = df["release_year"].value_counts().sort_index()

year_counts.plot(kind="line")

plt.title("Netflix Content by Release Year")
plt.xlabel("Release Year")
plt.ylabel("Number of Titles")

plt.show()
```
