Step 4 — Transform Categorical & Date Columns

```python
df["type"] = df["type"].astype("category")
```

```python
df["rating"] = df["rating"].astype("category")
```

```python
print(df["country"].dtype)
```

```python
df["date_added"] = pd.to_datetime(
    df["date_added"],
    errors="coerce"
)
```
